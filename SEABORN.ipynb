```python
import seaborn as sns
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5]
marks = [20, 40, 50, 70, 90]

sns.lineplot(x=hours, y=marks, marker="o")

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

students = ["Arun", "Bala", "Chitra", "Divya"]
marks = [70, 80, 65, 90]

sns.barplot(x=students, y=marks)

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Study_Hours": [1, 2, 3, 4, 5],
    "Marks": [20, 35, 50, 65, 80],
    "Attendance": [60, 70, 75, 85, 95]
}

df = pd.DataFrame(data)

correlation = df.corr()

sns.heatmap(correlation, annot=True)

plt.title("Correlation Heatmap")
plt.show()
```
