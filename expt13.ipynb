```python
from collections import Counter
import re

# Sample text
text = """Python is a popular programming language.
Python is used for data analysis and data science.
Data analysis helps to understand data.
Python makes data analysis easy and useful."""

# Convert text to lowercase
text = text.lower()

# Extract words
words = re.findall(r'\b\w+\b', text)

# Calculate frequency
frequency = Counter(words)

# Display output
print("Word Frequency Distribution:")
for word, count in frequency.items():
    print(word, ":", count)
```
