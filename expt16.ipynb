```python
import pandas as pd

data = {
    'reviewText': [
        'Good product and good quality',
        'Product quality is excellent',
        'Good product and useful'
    ]
}

df = pd.DataFrame(data)

review = df['reviewText'].str.lower().str.split().explode().value_counts()

print("Word Frequency Distribution:")
print(review)
```
