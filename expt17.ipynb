```python
import pandas as pd
import matplotlib.pyplot as plt

# Read CSV file
df = pd.read_csv('/content/creviews.csv', encoding='cp1252')

# Select required fields
data = df[['Age', 'Rating', 'Department Name', 'Review Text']]

# Display first 5 records
print("Customer Review Data:")
print(data.head())

# Calculate word frequency
review = data['Review Text'].dropna().str.lower().str.split().explode().value_counts()

# Display word frequency
print("\nWord Frequency Distribution:")
print(review)

# Find top 5 frequent words
rem_w = review.nlargest(5)

print("\nTop 5 Most Frequent Words:")
print(rem_w)

# Create bar plot
plt.bar(rem_w.index, rem_w.values)

plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title("Most Frequent Words")

plt.show()
```
